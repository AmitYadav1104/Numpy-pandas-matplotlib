```python
#create dashboard

import matplotlib.pyplot as plt
import numpy as np

# Data
students = ["Amit", "Rahul", "Priya", "Neha", "Riya"]

math = [85, 72, 90, 68, 78]
science = [80, 75, 88, 70, 82]
english = [78, 80, 92, 74, 76]
attendance = [95, 88, 98, 85, 90]

fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# ---------------- Line Plot ----------------
ax[0][0].plot(students, math, marker='o', label='Math')
ax[0][0].plot(students, science, marker='s', label='Science')
ax[0][0].plot(students, english, marker='^', label='English')

ax[0][0].set_title("Subject-wise Marks Comparison")
ax[0][0].set_xlabel("Students")
ax[0][0].set_ylabel("Marks")
ax[0][0].legend()
ax[0][0].grid(True)

# ---------------- Grouped Bar Chart ----------------
x = np.arange(len(students))
width = 0.25

ax[0][1].bar(x - width, math, width, label="Math")
ax[0][1].bar(x, science, width, label="Science")
ax[0][1].bar(x + width, english, width, label="English")

ax[0][1].set_xticks(x)
ax[0][1].set_xticklabels(students)
ax[0][1].set_title("Marks Comparison")
ax[0][1].set_xlabel("Students")
ax[0][1].set_ylabel("Marks")
ax[0][1].legend()

# ---------------- Pie Chart ----------------
ax[1][0].pie(
    attendance,
    labels=students,
    autopct="%1.1f%%",
    startangle=90
)
ax[1][0].set_title("Attendance Distribution")

# ---------------- Scatter Plot ----------------
ax[1][1].scatter(science, english, color="purple", s=100)

# Display student names beside points
for i in range(len(students)):
    ax[1][1].text(science[i] + 0.5, english[i], students[i])

ax[1][1].set_title("Science vs English")
ax[1][1].set_xlabel("Science Marks")
ax[1][1].set_ylabel("English Marks")
ax[1][1].grid(True)

# Dashboard Title
fig.suptitle("Student Performance Dashboard", fontsize=16)

plt.tight_layout()

plt.show()
```
