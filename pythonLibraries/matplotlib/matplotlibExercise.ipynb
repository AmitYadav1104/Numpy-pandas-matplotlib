```python
#stack bar chart

students = ["Amit", "Rohit", "Neha", "Priya"]

math = [40, 45, 35, 50]
science = [30, 25, 40, 35]

plt.bar(students,math,label="math",width=0.4)
plt.bar(students,science, bottom=math,label="science",width=0.4)

plt.title("students marks")
plt.xlabel("students")
plt.ylabel("marks")
plt.legend()

plt.show()
```

```python
#Example 2: Three Stacked Bars

subjects = ["Math", "Science", "English"]

A = [30, 25, 35]
B = [20, 30, 15]
C = [10, 15, 20]       #i am converting list into np.array


A=np.array(A)
B=np.array(B)
C=np.array(C)

plt.bar(subjects,A,label="A")
plt.bar(subjects,B,bottom=A,label="B")
plt.bar(subjects,C,bottom=A+B,label="C")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

subjects = ["Math", "Science", "English"]

# Class A
boys_A = [30, 25, 35]
girls_A = [20, 15, 25]

# Class B
boys_B = [25, 30, 20]
girls_B = [15, 20, 30]

x=np.arange(len(subjects))
width=0.35

plt.figure(figsize=(8,5))

#for left bar
plt.bar(x - width/2, boys_A,width, label="A Boys")
plt.bar(x - width/2, girls_A,width,  bottom=boys_A, label="A Girls")

#for right

plt.bar(x + width/2, boys_B,width, label="B Boys")
plt.bar(x + width/2, girls_B,width,  bottom=boys_B, label="B Girls")

plt.xticks(x,subjects)
plt.xlabel("Subjects")
plt.ylabel("Students")
plt.title("Side-by-Side Stacked Bar Chart")
plt.legend()
plt.show()
```

```python
#pie chart

marks = [40, 30, 20, 10]
subjects = ["Math", "Science", "English", "Computer"]

explode=[0.1,0,0,0]
plt.pie(marks,labels=subjects,autopct="%1.1f%%",colors=["red","blue","yellow","orange"],
explode = explode)

plt.title("subjects marks")
plt.show()
```

```python
plt.pie(marks,labels=subjects,startangle=90,shadow=True)

plt.axis("equal")
plt.show()
```

```python
sales = [50, 25, 15, 10]
products = ["Laptop", "Mobile", "Tablet", "Accessories"]

explode=[0.1,0,0,0]
plt.pie(
    sales,
    labels=products,
    colors=["red","green","black","purple"],
    autopct="%1.1f%%",
    shadow=True,
    startangle=90,
    explode=explode,
)

plt.axis("equal")
plt.show()
```

```python
#advanced matplotlib

x=np.arange(1,6)

y1=x
y2=x**2
y3=x**3
y4=np.sqrt(x)

fig,ax=plt.subplots(2,2,figsize=(8,6))

ax[0][0].plot(x,y1,color="purple",marker="o")
ax[0][0].set_title("Linear")

ax[0][1].plot(x,y2,color="red",marker="o")
ax[0][1].set_title("square")

ax[1][0].plot(x,y3,color="blue",marker="o")
ax[1][0].set_title("cube")

ax[1][1].plot(x,y4,color="green",marker="o")
ax[1][1].set_title("sqrt")

plt.grid(True)
plt.tight_layout()
plt.show()
plt.style.use("ggplot")
```
