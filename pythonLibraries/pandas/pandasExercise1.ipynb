```python
#series from list
import numpy as np
import pandas as pd

#string

country=['india','pakistan','us','uk','china']

pd.Series(country)


#integer

marks=[54,65,69,85,95]
pd.Series(marks)


#custom index

marks=[54,65,69,85,95]
sub=['english','maths','science','geo','history']

pd.Series(marks, index=sub,dtype=np.int32)

#setting names

pd.Series(marks, index=sub, name='amit')
```

```python
#series from dict

marks={
    "eng":55,
    "maths":65,
    "sci":70,
    "geo":92,
    "history":95
}

pd.Series(marks)
```
