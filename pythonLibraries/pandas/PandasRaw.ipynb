```python
import pandas as pd

x = {
    "Name": ["Amit", "Rohit", "Amit", "Neha", "Rohit", "Neha", "Amit", "Rohit"],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance", "IT", "HR"],
    "Salary": [50000, 45000, 55000, 60000, 48000, 62000, 58000, 50000],
    "Age": [22, 25, 23, 30, 26, 31, 24, 27]
}

df = pd.DataFrame(x)
df

#Easy Level
#1.
#Group by Name and find the total salary.

df.groupby("Name")["Salary"].sum()

#2.
#Group by Department and find the average salary.

df.groupby("Department")["Salary"].mean()

#3.

#Count how many employees belong to each city.

df["City"].value_counts()
df.groupby("City")["Name"].count()

#4.

#Find the maximum age in each department.
df.groupby("Department")["Age"].max()


#Find the maximum age in each department with name

df.loc[df.groupby("Department")["Age"].idxmax(),["Department","Name","Age"]]

#5.

#Find the minimum salary for each city.
df.groupby("City")["Salary"].min()


#6.

#Find the number of employees in each department.
df.groupby("Department")["Name"].count()

#7.

#Find the average age of employees in each city.

df.groupby("City")["Age"].mean()

#8.

#Find the highest salary earned by each employee.
df.groupby('Name')["Salary"].max()

#9.

#Find the lowest age for each department.
df.groupby("Department")["Age"].min()

#Find the lowest age for each department with name.
df.loc[df.groupby("Department")["Age"].idxmin(),["Name","Department","Age"]]

#10.

#Find salary sum city-wise.

df.groupby("City")["Salary"].sum()
```
