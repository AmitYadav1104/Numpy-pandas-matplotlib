```python
#np.sort() (5 Questions)
#Sort [5,2,8,1,9].
np.sort([5,2,8,1,9])
#Sort an array in descending order.
arr=[5,2,8,1,9]
arr1=np.sort([arr])[::-1]
arr1
#Sort each row of a 3×3 matrix.
a=np.array([5,6,1,2,8,4,3,9,7]).reshape(3,3)
a
b=np.sort(a, axis=0)
#print(a)
#print(b)

#Sort each column of a 3×3 matrix.
c=np.sort(a, axis=1)
#print(a)
#print(c)
#Sort a random array of 20 numbers

A=np.random.random(20)
B=np.sort(A)
print(A)
print(B)
```

```python
#np.where() (5 Questions)
#Replace negative numbers with 0.
A=np.array([5,2,-2,4,-3,6,1])
np.where(A<0,0,A)

#Replace numbers greater than 50 with 100.
B=np.array([150,200,45,50,35,101])
np.where(B>50,100,B)
#Label numbers as "Even" or "Odd".
A=np.array([5,2,-2,4,-3,6,1])
np.where(A%2==0,'even','odd')
#Replace all values less than 10 with -1.
C=np.array([10,5,12,8,14,9])
np.where(C<10,-1,C)
#Convert marks above 40 to "Pass" and others to "Fail"
D=np.array([45,50,35,55,39,65])
np.where(D>40,"pass","fail")
```

```python
#7. Boolean Indexing (5 Questions)
#Extract numbers greater than 50.
D=np.array([45,50,35,55,39,65])
D[D>50]
#Extract all even numbers.
A=np.array([5,6,7,8,2,4,9])
(A[A%2==0 ]).dtype

#Extract all negative numbers.
A=np.array([5,2,-2,4,-3,6,1])
A[A<0]
#Extract values between 20 and 50.
D=np.array([45,50,35,55,39,65])
D[(D>20) & (D<50)]
#Extract elements divisible by both 2 and 3.
B=np.array([45,50,35,55,39,65,52,64,68,66])
print(B[(B%2==0) & (B%3==0)])
```

```python
#np.concatenate() ( Questions)
#Join two 1D arrays.
a=np.arange(7)
b=np.arange(7,14)
np.concatenate((a,b))

#Join three arrays.
c=np.arange(14,20)
np.concatenate((a,b,c)).reshape(4,5)
#Join two 2×2 matrices row-wise.
x=np.arange(4).reshape(2,2)
y=np.arange(5,9).reshape(2,2)
np.concatenate((x,y), axis=0)
#Join two 2×2 matrices column-wise.
x=np.arange(4).reshape(2,2)
y=np.arange(5,9).reshape(2,2)
np.concatenate((x,y), axis=1)
#Join two random arrays of size 5.
a=np.random.random(5)
b=np.random.random(5)
np.concatenate((a,b))

x=np.random.randint(1,100,5)
y=np.random.randint(1,100,5)
np.concatenate((x,y))

#uniform
d=np.random.uniform(1,20,5)
c=np.random.uniform(1,20,5)
np.concatenate((d,c)).reshape(2,5)
```
